```python
import pandas as pd
import requests
import io
from src.utils import load_data_from_google_drive

# Define the base URLs for the yellow and green taxi data
base_url_yellow = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{year}-{month}.parquet"
base_url_green = "https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_{year}-{month}.parquet"
zones_url = 'https://drive.google.com/file/d/12VgjWXkyEBsxzuKFxIkAevEbao85ei0T/view?usp=sharing'

# Define the months and year you're interested in
months = ['01', '02']
year = '2019'

# Headers to mimic a browser request (helps with WAF/bot protection)
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
    'Accept': 'application/octet-stream,*/*',
    'Accept-Language': 'en-US,en;q=0.9',
}

def download_parquet(url, headers):
    """Download parquet file with proper headers and error handling."""
    response = requests.get(url, headers=headers, timeout=60)
    response.raise_for_status()
    
    if len(response.content) == 0:
        raise ValueError(f"Empty file received from {url}")
    
    return pd.read_parquet(io.BytesIO(response.content))

# Create empty lists to store the dataframes
df_yellow_list = []
df_green_list = []

# Loop over the months
for month in months:
    url_yellow = base_url_yellow.format(year=year, month=month)
    url_green = base_url_green.format(year=year, month=month)
    
    try:
        print(f"Downloading yellow taxi data for {year}-{month}...")
        dfy = download_parquet(url_yellow, headers)
        df_yellow_list.append(dfy)
        print(f"  ✓ Downloaded {len(dfy):,} rows")
    except Exception as e:
        print(f"  ✗ Failed to download yellow taxi {year}-{month}: {e}")
    
    try:
        print(f"Downloading green taxi data for {year}-{month}...")
        dfg = download_parquet(url_green, headers)
        df_green_list.append(dfg)
        print(f"  ✓ Downloaded {len(dfg):,} rows")
    except Exception as e:
        print(f"  ✗ Failed to download green taxi {year}-{month}: {e}")

# Concatenate the list of dataframes into a single dataframe
if df_yellow_list:
    df_yellow = pd.concat(df_yellow_list, ignore_index=True)
    print(f"\nTotal yellow taxi records: {len(df_yellow):,}")
else:
    raise RuntimeError("No yellow taxi data could be downloaded. See errors above.")

if df_green_list:
    df_green = pd.concat(df_green_list, ignore_index=True)
    print(f"Total green taxi records: {len(df_green):,}")
else:
    raise RuntimeError("No green taxi data could be downloaded. See errors above.")

df_zones = load_data_from_google_drive(url=zones_url)
print(f"Zones loaded: {len(df_zones)} zones")
```

To create a proper forecast for both the yellow and green taxis a time-based splitting approach has been used. ##### Forecast evaluation – Yellow taxis
The forecast is plotted above, both for the whole period of two years, and a zoomed in view of the last three months. The forecast is the orange line, plotted with its confidence interval, while the actual data, the forecast is tested on is plotted as the blue line. 

The forecast is evaluated with three different evalutaion metrics: MAE, RMSE and MAPE.

##### MAE and RMSE:
The MAE and RMSE are both great metrics to understand how close or far of the forecast generelly are from the acutal data. MAE shows the average error pr day, which is on 22.870 rides. This may sound like a lot, but when there are between 100.000 and over 250.000 rides pr. day, it is actually quite acceptable. The RMSE works as a great supplement, as it squares the errors, so it 'punishes' few larger errors more than many small errors, where as it is all the same for the MAE. In this case the RMSE is relevant because it would be a major crisis for the taxi company if the forecast didn't take a holiday like thanksgiving og new years into account and miscalculated with 100.000 rides on one single day, compared to miscalculating around 10.000 rides on different normal days. The RSME is on 27.399 errors per day, which is not too far from the MAE, and the errors is more or less evenly distributed, which is good.

Even though these to metrics says a lot about this forecast, it can be somewhat hard to determine how good or bad an average error of 22.000 are for this forecast. These metrics are also irrelevant when it is needed to compare one forecast to another, ex. an MAE of 10.000 is amazing for the yellow taxis, it would be catastrofical for the green taxis, since they generally have a way smaller amount of rides. To compare forecasts, the MAPE is a great evaluation metric:

##### MAPE:
The MAPE shows the relative error of the forecast. Sometimes it can be difficult to know exactly what the impact of an error of 20.000 rides per day really means, or how much better one forecast is compared to another, and that is why the MAPE is great, since it determines the relative error of the forecast and returns how many percent the forecast is off from the actual data. The forecast for the yellow taxis has a MAPE on 10,53% which is a very reasonable MAPE for a forecast of this size.
