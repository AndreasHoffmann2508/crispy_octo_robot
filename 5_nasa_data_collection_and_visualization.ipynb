```python
# Set the start and end dates for the data you want to fetch
start_date = datetime.strptime('2022-01-01', '%Y-%m-%d')
end_date = start_date + timedelta(days=365)  # 1 year later

# Initialize a list to store the data
data = []

# Fetch data from the NASA API 7 days at a time
# The introduction of the API is on https://api.nasa.gov, under "Browse APIs" -> "Asteroids NeoWs"
# You can look into the example query in the link below to see what the data look like:
# https://api.nasa.gov/neo/rest/v1/feed?start_date=2015-09-07&end_date=2015-09-08&api_key=DEMO_KEY
current_date = start_date
while current_date < end_date:
    next_date = min(current_date + timedelta(days=7), end_date)
    response = requests.get(f'https://api.nasa.gov/neo/rest/v1/feed?start_date={current_date.strftime("%Y-%m-%d")}&end_date={next_date.strftime("%Y-%m-%d")}&api_key={api_key}')
    data.append(response.json())
    current_date = next_date
    time.sleep(1)  # To avoid hitting the rate limit

# Now 'data' contains the NEO data for the 1-year period
```

**Choose an appropriate display (Chapter 2)**
- *Number of NEOs per week* is a count that changes over time, so a single series **line plot** shows the trend and continuity best. Having each week as the time points adds cluster, as it becomes overcrowded and hard to depict what week corresponds to the respective point on the line graph. To avoid this, showing every 3rd week makes the visualization easy to read and maintains a good data-ink ratio.
- *Distribution of NEO sizes* is the shape of a single variable, so a **histogram** is the obvious choice for showing the distribution. A **log x-axis** is used because the sizes span several orders of magnitude (a few huge asteroids would otherwise cram everything into the first bin).
- *Average NEO size per week* compares a summary value across discrete time buckets, so a **bar plot** lets you read those week-to-week comparisons precisely.
- The *weekday x week-of-year* counts form a 2-D grid, so a **heat map** encodes the whole table compactly with colour intensity.

**Eliminate clutter (Chapter 3)**
- A single accent colour, a white background, and removed top/right spines (`sns.despine`) keep the focus on the data.
- The crowded weekly tick labels are hidden on the bar chart (there are too many to read), and the heat map drops gridlines so the colour field stays uninterrupted.
- Each chart shows one series with a plain, left-aligned title, so no legend is needed.
